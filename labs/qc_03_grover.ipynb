### Theory recap (Grover 1996; N&C §6.1)
Let $\sin\theta = \sqrt{M/N}$. In the plane spanned by $\lvert\text{good}\rangle$ and $\lvert\text{bad}\rangle$, the start state is $\lvert s\rangle = \sin\theta\lvert\text{good}\rangle + \cos\theta\lvert\text{bad}\rangle$ and one iteration $G = DO$ is a rotation by $2\theta$. After $k$ iterations
$$P_{\text{success}}(k) = \sin^2\big((2k+1)\theta\big), \qquad k^* = \Big\lfloor \frac{\pi}{4}\sqrt{N/M}\Big\rfloor.$$
Overshooting **decreases** the probability — the algorithm is periodic.

Construction: $D = H^{\otimes n}(2\lvert0\rangle\langle0\rvert - I)H^{\otimes n}$ and $2\lvert0\rangle\langle0\rvert - I = -X^{\otimes n}(\text{MCZ})X^{\otimes n}$ (global phase $-1$ is irrelevant). MCZ $= H_{n-1}\,\text{MCX}\,H_{n-1}$ (`qc.mcx(controls, target)`).
Marked items are integers in **Qiskit's little-endian** indexing: item $x$ has bit $q$ = `(x >> q) & 1`.
